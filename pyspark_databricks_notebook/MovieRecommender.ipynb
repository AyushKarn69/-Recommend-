# Movie Recommender

PySpark version of the movie recommendation system.

```python
from pyspark.mllib.recommendation import ALS, Rating
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("movie-recommender").getOrCreate()
sc = spark.sparkContext
```

```python
ratings_path = "hdfs://localhost:9000/movie/u.data"
items_path = "hdfs://localhost:9000/movie/u.item"

raw_ratings = sc.textFile(ratings_path)
ratings = raw_ratings.map(lambda line: line.split("\t")) \
    .map(lambda fields: Rating(int(fields[0]), int(fields[1]), float(fields[2])))

model = ALS.train(ratings, rank=10, iterations=10, lambda_=0.01)

films = model.recommendProducts(100, 5)
rating = model.predict(100, 1311)
users = model.recommendUsers(200, 5)

print(films)
print(rating)
print(users)
```
